```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
sent
```

```python
### One Hot Representation
one_hot_repr=[one_hot(words,voc_size)for words in sent]
one_hot_repr
```

```python
## word Embedding Representation

from tensorflow.keras.layers import Embedding
#from tensorflow.keras.processing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(embedded_docs)
```

```python
## feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

```python
embedded_docs[0]
```

```python
model.predict(embedded_docs[0])
```

```python
import numpy as np

# Assuming embedded_docs[0] is a 1D NumPy array
single_sample = embedded_docs[0]

# Reshape the 1D array (n_features,) into a 2D array (1, n_features)
# The '-1' in reshape tells NumPy to automatically calculate the dimension size.
reshaped_sample = single_sample.reshape(1, -1)

# Now, make the prediction
prediction = model.predict(reshaped_sample)

print(prediction)
```

```python
# Pass the sample inside a list to create the "batch" dimension
prediction = model.predict([embedded_docs[0]])

print(prediction)
```

```python
import numpy as np

# Assuming embedded_docs[0] is a 1D NumPy array
single_sample = embedded_docs[0]

# Add a new axis at the beginning (axis=0)
reshaped_sample = np.expand_dims(single_sample, axis=0)

# Now, make the prediction
prediction = model.predict(reshaped_sample)

print(prediction)
```
